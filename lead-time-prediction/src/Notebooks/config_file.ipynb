```python
# ── Environment detection ─────────────────────────────────────────────────────
if spark.conf.get("spark.databricks.clusterUsageTags.clusterOwnerOrgId") == "7181820732839861":
    env = "d"    # DEV / TEST workspace
elif spark.conf.get("spark.databricks.clusterUsageTags.clusterOwnerOrgId") == "4796304964978539":
    env = "a"    # ACC workspace
elif spark.conf.get("spark.databricks.clusterUsageTags.clusterOwnerOrgId") == "4844910598668398":
    env = "p"    # PRD workspace
else:
    env = "d"    # fallback to dev

# ── Catalog ───────────────────────────────────────────────────────────────────
UNITY_CATALOG = f"automation_{env}"   # automation_d | automation_a | automation_p

# ── Medallion layer schemas ───────────────────────────────────────────────────
BRONZE = f"{UNITY_CATALOG}.bronze"    # raw data from Synapse
SILVER = f"{UNITY_CATALOG}.silver"    # cleaned & enriched
GOLD   = f"{UNITY_CATALOG}.gold"      # ML / statistical outputs

# ── Ensure schemas exist ──────────────────────────────────────────────────────
for _s in [BRONZE, SILVER, GOLD]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {_s}")

# ── Synapse catalog ───────────────────────────────────────────────────────────
SYNAPSE_CATALOG_MAP = {"d": "`synapse-prd`", "a": "`synapse-acc`", "p": "`synapse-prd`"}
SYNAPSE_CATALOG = SYNAPSE_CATALOG_MAP[env]
SYNAPSE_SCHEMA  = "pbi"

print(f"env={env}  catalog={UNITY_CATALOG}  synapse={SYNAPSE_CATALOG}")
print(f"  bronze={BRONZE}  silver={SILVER}  gold={GOLD}")
```
